# Problem: Variational Dropout in RNNs for Language Modeling

## Problem Statement
You are tasked with implementing **Variational Dropout** (also known as locked dropout) for an LSTM network on a language modeling task. Unlike naive dropout which samples a new dropout mask at each timestep, variational dropout uses the **same dropout mask across all timesteps** in a sequence, which has been shown to improve regularization in RNNs.

### Objectives
1. Implement a custom LSTM with **Variational Dropout** (locked dropout across time steps)
2. Implement a baseline LSTM with **Naive Dropout** (standard dropout at each timestep)
3. Train both models on a language modeling dataset (Penn Treebank)
4. Compare the performance and training stability of both approaches
5. Visualize training curves to demonstrate the difference

### Requirements
- Implement `VariationalDropout` layer that maintains the same mask across timesteps
- Create two LSTM models: one with variational dropout, one with naive dropout
- Use perplexity as the evaluation metric
- Visualize and compare training/validation loss curves

### Dataset
**Penn Treebank (PTB)**: A standard benchmark for language modeling containing Wall Street Journal articles.

<details>
  <summary>💡 Hint</summary>
  
  **Variational Dropout**: Sample the dropout mask once per sequence and apply it at every timestep. This can be implemented by generating a mask of shape `(batch_size, hidden_size)` and reusing it.
  
  **Naive Dropout**: Use standard PyTorch dropout which samples a new mask at each forward pass.
  
  **Key Difference**: Variational dropout prevents the network from learning to rely on specific timestep patterns, leading to better generalization.
